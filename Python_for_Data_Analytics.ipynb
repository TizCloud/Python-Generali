##While Loop

```python
count = 0

while (count < 10):
  print ('This count is: ', count)
  count = count + 1

print('The End')
```

```python
count = 0

while count < 9 :
  print (count, "is less than 9.")
  count = count + 1
else:
  print (count, "is not less than 9.")
```

```python
count = 1
sum = 0.0

print ("Exit this program, please type 0 or 0.0 :")

mynum = float(input("Enter a number : "))

while mynum != 0 or mynum != 0.0:
  sum += mynum  # Sum = sum + mynum
  avg = sum / count
  count += 1    # count = count + 1
  print("Average of number is : ", avg)
  mynum = float(input("Enter a number : "))
```

```python
for chr in 'Python':
  if chr == 'h':
    break
  print('Current Character :', chr)

print ("--------------------------------")

myvar = 10
while myvar > 0 :
  print ('Current variable value :', myvar)
  myvar = myvar - 1
  if myvar == 5:
    break
```
