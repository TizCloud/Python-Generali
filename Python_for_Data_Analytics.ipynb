##if statements

```python
myname = input ("Enter name :")
price = float(input("Enter price of %s :" %myname))

if price >= 1000:
  temp = price * 0.05
  price = price - temp

vat = price * 0.07
price = price + vat

print("The price of %s (inc VAT 7%%) is %.2f %s and VAT %.2f %s" %(myname, price, "Baht.", vat , "Baht." ))
```

```python
myscore = float(input("Enter your score : "))
mymsg = "Your grade is "

if myscore >= 80:
  print(mymsg + "A")
elif myscore >= 70:
  print(mymsg + "B")
elif myscore >= 60:
  print(mymsg + "C")
elif myscore >= 50:
  print(mymsg + "D")
else
  print(mymsg + "F")

```

```python
count = 1
sum = 0.0

print ("Exit this program, please type 0 or 0.0 :")

mynum = float(input("Enter a number : "))

while mynum != 0 or mynum != 0.0:
  sum += mynum  # Sum = sum + mynum
  avg = sum / count
  count += 1    # count = count + 1
  print("Average of number is : ", avg)
  mynum = float(input("Enter a number : "))
```
